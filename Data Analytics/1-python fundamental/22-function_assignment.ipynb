```python
# Define a function that takes a variable number of integer arguments and returns their product. Test the function with different inputs.
def multiply(*args):
    product = 1
    for i in args:
        product *=i 
    return product
print(multiply(10,2,5,4,3,6))
print(multiply(8,1,2,3,2,5,7))
print(multiply(13,5,4,6,8,7,0))
print(multiply(2,3,4))
```

```python
# Define a function that contains another function inside it. The outer function should take two integers as input and return the 
# result of the inner function, which multiplies the two integers. Test the function with different inputs.

def outer(a,b):
    def inner(x,y):
        return x*y
    return inner(a,b)
print(outer(10,8))
print(outer(19,8))
print(outer(11,8))
```

```python
# Define a function that takes a single integer as input and returns the integer squared, cubed, and raised to the power of four. 
# Test the function with different inputs.

def power(a):
    return a**2,a**3,a**4
print(power(2))
print(power(23))
print(power(12))
```

```python
# Define a recursive function that calculates the factorial of a given number. Test the function with different inputs.
def factorial(n):
    if n==0:
        return 1
    return n*factorial(n-1)
print(factorial(5))
print(factorial(51))
print(factorial(15))
```

```python
# Define a decorator function that prints 'Executing function...' before executing a function and 'Function executed.' after executing it. 
# Apply this decorator to a function that takes a list of integers and returns their sum. Test the decorated function with different lists.
import time

# Decorator
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Executing function...")
        
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()

        print("Function executed.")
        print(f"Time taken: {end - start:.6f} seconds")
        
        return result
    return wrapper


# Function using decorator
@my_decorator
def multiply_list(numbers):
    result = 1
    for num in numbers:
        result *= num
    return result


# Test cases
print("Result:", multiply_list([1, 2, 3, 4]))
print("Result:", multiply_list([5, 6, 7]))
```

```python
# Define a function that takes variable-length arguments and keyword arguments and prints them. Test the function with different inputs.
def print_args_kwargs(*args,**kwargs):
    print("*args",args)
    print("*kwargs",kwargs)
 
print_args_kwargs(1, 2, 3, a='apple', b='banana')
print_args_kwargs('hello', 'world', x=10, y=20)
```

```python
# Define a higher-order function that takes a function and a list of integers as arguments, and applies the function to each integer in the list.
#  Test with different functions and lists.

def higer_order_function(func,lst):
    return [func(x) for x in lst]

print(higer_order_function(lambda x:x*2,[1,2,3,4,5]))
print(higer_order_function(lambda x:x+2,[11,22,33,44,55]))
```
