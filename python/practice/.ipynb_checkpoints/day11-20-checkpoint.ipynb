```python
# Write a program that finds the minimum, maximum, and average of a list of numbers without using built-in min/max.
li = [2,63,4,54]
max=li[0]
min=li[0]
sum = 0
for i in li:
    if max<i:
        max=i
    if min>i:
        min=i
    sum += i
avg = sum/len(li)
print("max: ",max)
print("min: ",min)
print("Avg: ",avg)
```

```python
# What are dict.keys(), dict.values(), and dict.items()? Write a loop using each.
# In Python, these three methods allow you to access the different components of a dictionary:dict.keys(): Returns a view of all the names (keys) in the dictionary.dict.values(): Returns a view of all the data (values) stored in the dictionary.dict.items(): Returns a view of both as pairs (tuples).
dic ={
    "name":"Abdullah",
    "age":26,
    "gender":"Male"
}
# 1. Looping through keys
for key in dic.keys():
    print(f"Key: {key}")

# 2. Looping through values
for value in dic.values():
    print(f"Value: {value}")

# 3. Looping through items (most common for both)
for key, value in dic.items():
    print(key," : ",value)
```

```python
# Write a dictionary comprehension that maps numbers 1-10 to their squares
sq = {x : x**2 for x in range(1,11)}
for key,value in sq.items():
    print(key," : ",value)
```

```python
# How do you merge two dictionaries in Python? Show at least two methods including the ** operator.
dic1 ={
    "name":"Abdullah",
    "age":26,
    "gender":"Male"
}
dic2 ={
    "userName":"Tayyab",
    "skill":"Data analysts",
    "city":"kasur"
}
merged = {**dic,**dic2}
print("using ** operator: ",merged)

merged = dic1 | dic2
print("using | operator: ",merged)

dic1.update(dic2)
print("using .update() operator: ",dic1)

dic1 |= dic2
print("using |= operator: ",dic1)
```

```python
# Write a program that counts the frequency of each word in a sentence using a dictionary
def countWord(str):
    sentence = str.lower()
    words = sentence.split()
    dic = {}
    for word in words:
        word = word.strip('.,!?;:"()')
        if word:
            if word in dic:
                dic[word] +=1
            else:
                dic[word] =1
    return dic
str = "The quick brown fox jumps over the lazy dog. The dog barked!"
print(countWord(str))
```

```python
# Write a program that counts the frequency of each character in a sentence using a dictionary

def countCh(str):
    sentence = str.lower()
    dic = {}
    for ch in sentence:
        if ch in dic:
            dic[ch] +=1
        else:
            dic[ch] =1
    return dic
str = "The quick brown fox jumps over the lazy dog. The dog barked!"
counted = countCh(str)
for key,value in counted.items():
    print(key," : ",value)
```

```python
# What is a nested dictionary? Create one representing a school with classes and students
university = {
    "class9":{
        "strength":20,
        "preent":17,
        "lecture":{
            "english":"9Am",
            "math":"10Am",
            "compuer":"11Am",
        }
    },
    "class10":{
        "strength":18,
        "present":13,
        "lecture":{
            "english":"12Am",
            "math":"01Am",
            "compuer":"02Am",
        }
    }
}
for class_name,class_info in university.items():
    print("Class: ",class_name)
    for key,value in class_info.items():
        if key=="lecture":
            print("Lecture: ")
            for lec,time in value.items():
                print(lec," : ",time)
        else:
            print(key," : ",value)
    print()
```

```python
# Write a program that groups a list of words by their first letter using a dictionary.
li = ["apple","banana","mango","apricot","blueberry","avacado"]
dic = {}
for word in li:
    letter = word[0]
    if letter not in dic:
        dic[letter]=[]
    dic[letter].append(word)
for key,value in dic.items():
    print(key," : ",value)
```

```python
# What is defaultdict from the collections module? Rewrite the word-frequency program using it.
from collections import defaultdict
li = ["apple","banana","mango","apricot","blueberry","avacado","apple"]
dic = defaultdict(int)
for i in li:
    dic[i] +=1
for key,value in dic.items():
    print(key," : ",value)
```

```python
# Write code to split a sentence into a list of words, then join them back with a hyphen
def stringFormat(str1):
    str1 = str1.split()
    text = ""
    for i in str1:
        if i != str1[-1]:
            text += i
            text += "-"
        else:
            text += i
    return text
text = "Abdullah Tayyab @ gamil .com"
print(stringFormat(text))

# or 

text = text.split()
text = "-".join(text)
print(text)
```
