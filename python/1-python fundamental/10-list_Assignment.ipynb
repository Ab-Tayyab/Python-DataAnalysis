```python
# Create a list of dictionaries where each dictionary represents a student with keys 'name' and 'score'. Sort the list of dictionaries by the 'score' in descending order and print the sorted list.
student_list = [
    {"name": "Alice", "score": 88},
    {"name": "Bob", "score": 72},
    {"name": "Charlie", "score": 95},
    {"name": "David", "score": 65},
    {"name": "Eve", "score": 78},
]
reverse_sorted = sorted(student_list,key=lambda x:x["score"],reverse=True)
print("Reverse sorted list")
for i in reverse_sorted:
    print(i)

# or 

def sort_li(student_list):
    return student_list["score"]
reverse_sort = sorted(student_list,key=sort_li,reverse=True)
print("Reverse sorted list")
for i in reverse_sort:
    print(i)
```

```python
# Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.
def tranposed_matrix(a):
    # tranposed = [[a[j][i] for j in range(len(a))] for i in range(len(a[0]))]
    # return tranposed

    # or 
    tranposed = []
    for i in range(len(a[0])):
        row=[]
        for j in range(len(a)):
            row.append(a[j][i])
        tranposed.append(row)
    return tranposed


matrix = [[1, 2, 3], 
          [4, 5, 6], 
          [7, 8, 9]]
tranposed = tranposed_matrix(matrix)
print("Original Matrix")
for row in matrix:
    print(row)
print("Tranpose Matrix")  
for row in tranposed: 
    print(row)
```

```python
# Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
flattens_list=[]
for i in range(len(matrix)):
    for j in range(len(matrix[0])):
        flattens_list.append(matrix[i][j])
print("Original list")
print(matrix)
print("Flattens list")
print(flattens_list)
```

```python
# Write a function that rotates a list by n positions. Print the original and rotated lists.
def rotate_list(li,n):
    return li[n:]+li[:n]

li=[1,2,3,4,5]
print(li)
print("Rotate list: ", rotate_list(li,2))
```
