```python
# What is short-circuit evaluation in Python? Demonstrate with 'and' and 'or'
# Short-circuit evaluation is a functional optimization in Python where the interpreter stops evaluating a boolean expression as soon as its final truth value can be determined. This technique, also known as lazy evaluation, improves performance and prevents errors by skipping unnecessary computations.
# The second part (1/0) is never evaluated because the first is False.
result = False and (1 / 0 > 1) 
print(result)  # Output: False (no ZeroDivisionError)
# The second part is skipped because the first is True.
result = True or (1 / 0 > 1)
print(result)  # Output: True (no ZeroDivisionError)
```

```python
# What is the match-case statement (Python 3.10+)? Write an example that maps day numbers to day names.
# Introduced in Python 3.10, the match-case statement is a feature known as Structural Pattern Matching. It functions similarly to the switch-case statements found in other languages but is more powerful, allowing you to match complex data structures and extract values directly.
value = 6
match value:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Value!")
```
