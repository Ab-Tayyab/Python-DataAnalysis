```python
# What is function docstring? Write a function with a proper docstring and access it using __doc__.
# A function docstring (documentation string) is a special string literal placed as the first statement in a function's body. Unlike regular comments, docstrings are stored as metadata and are accessible at runtime using the __doc__ attribute or the help() function.
def area(r):
    """ Calculate the area of a circle
    Args:
         Radius: Radius of a circle
    Return:
        Float value as a return value"""
    import math
    return math.pi*(r**2)
print(area.__doc__)
```

```python
# What is the scope of a variable inside a function? Explain the LEGB rule with an example.
# The scope of a variable refers to where it can be accessed in a program.
# A variable defined inside a function has local scope, meaning it can only be used within that function.

def var():
    n=10
    print(n)
var()
# print(n)  # error

# 🔑 LEGB Rule in Python

# Python resolves variable names using the LEGB rule, which stands for:

# L – Local → Inside the current function
# E – Enclosing → Inside outer (nested) functions
# G – Global → Defined at the top level of the script
# B – Built-in → Predefined names (like print, len, etc.)

# Python searches for variables in this order: Local → Enclosing → Global → Built-in

x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print(x)   # Looks in Local first
    inner()
outer()

x = "global"
def outer():
    x = "enclosing"
    def inner():
        print(x)   # Not in local → finds in enclosing
    inner()
outer()
```

```python
# What are decorators in Python? Write a simple decorator that prints 'Before' and 'After' around a function call.
# A decorator is a function that wraps another function to extend or modify its behavior without changing its code.

def simple_decorator(func):
    def wrapper():
        print("before")
        func()
        print("after")
    return wrapper
@simple_decorator
def greet():
    print("Hello")
greet()
```

```python
# Write a decorator 'timer' that measures and prints the execution time of a function.
# Measures execution time of a function.
import time
def times(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        print(f"time is: {end-start}")
        return result
    return wrapper
@times
def cal():
    time.sleep(1)
cal()
```

```python
# What is functools.wraps and why should you use it in a decorator?

# wraps preserves metadata (like function name, docstring) of the original function.
# Without it:
# Function name becomes wrapper
# Docstring is lost

from functools import wraps

def decorator(func):
    @wraps(func)
    def wrapper(*args,**kwargs):
        return func(*args,**kwargs)
    return wrapper
```

```python
# Write a memoization decorator from scratch that caches results of expensive function calls.
# Caches results to avoid recomputation.
def memoize(func):
    cache = {}

    def wrapper(*args):
        if args in cache:
            return cache[args]
        result = func(*args)
        cache[args] = result
        return result

    return wrapper

@memoize
def expensive(n):
    print("Computing...")
    return n * n

print(expensive(4))
print(expensive(4))  # Cached
```

```python
# What is functools.lru_cache? Rewrite the Fibonacci function using it and compare performance.
# A built-in decorator for caching with a size limit.
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n-1) + fib(n-2)

print(fib(40))
# Performance:
# Without cache → exponential time
# With lru_cache → near linear time
```

```python
# Write a function that accepts a function and a list, and returns a new list with the function applied to each element (custom map).
def custom_map(func, lst):
    return [func(x) for x in lst]

print(custom_map(lambda x: x * 2, [1, 2, 3]))
```

```python
# Explain the difference between map(), filter(), and reduce(). Provide an example of each.

# map() → transforms elements
list(map(lambda x: x*2, [1,2,3]))
# [2, 4, 6]

# filter() → selects elements
list(filter(lambda x: x % 2 == 0, [1,2,3,4]))
# [2, 4]

# reduce() → reduces to single value
from functools import reduce

reduce(lambda x, y: x + y, [1,2,3,4])
# 10
```

```python
# What is a partial function (functools.partial)? Create a partial function for pow() that always squares a number
from functools import partial

square = partial(pow, 2)

print(square(5))  # 2^5 = 32 (note: order matters!)

# correct squaring
square = partial(pow, exp=2)
print(square(5))  # 25
```

```python
# Write a function pipeline: compose(f, g) that returns a function applying g then f.
def compose(f, g):
    return lambda x: f(g(x))

def add1(x):
    return x + 1

def double(x):
    return x * 2

h = compose(add1, double)
print(h(3))  # (3*2)+1 = 7
```

```python
# What is a generator function (using yield)? Write one that generates an infinite sequence of Fibonacci numbers.
def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

fib = fibonacci()

for _ in range(10):
    print(next(fib))
```
