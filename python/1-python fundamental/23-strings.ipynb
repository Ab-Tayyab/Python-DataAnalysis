```python
# Write code to split a sentence into a list of words, then join them back with a hyphen
def stringFormat(str1):
    str1 = str1.split()
    text = ""
    for i in str1:
        if i != str1[-1]:
            text += i
            text += "-"
        else:
            text += i
    return text
text = "Abdullah Tayyab @ gamil .com"
print(stringFormat(text))

# or 

text = text.split()
text = "-".join(text)
print(text)
```

```python
# Write a function that converts a camelCase string to snake_case using string methods
def camelcase_to_snake(text):
    str1 = ""
    for i in text:
        if i.isupper():
            str1 +="_"+i.lower()
        else:
            str1 += i.lower()
    return str1
text = ("letFindMe")
print(camelcase_to_snake(text))
```

```python
# Use str.translate() and str.maketrans() to remove all punctuation from a string
import string

text = "Hello, world! How are you?"

table = str.maketrans("", "", string.punctuation)
clean = text.translate(table)

print(clean)
```

```python
# Write a program that finds the longest common prefix of a list of strings.
def longest_prefix(strs):
    if not strs:
        return ""
    prefix = strs[0]
    for i in strs[1:]:
        while not i.startswith(prefix):
            prefix = prefix[:-1]
    return prefix
print(longest_prefix(["flower", "flow", "flight"]))
```

```python
# Write a function that formats a large number with commas (e.g., 1234567 → '1,234,567') without using f-string formatting.
def format_number(strs):
    s = str(strs)
    result = ""
    while len(s)>3:
        result += "," + s[-3:] + result
        s=s[:-3]
    return s+result
print(format_number(12345678))
```
