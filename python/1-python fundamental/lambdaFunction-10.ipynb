```python
# What is a lambda function in Python? Write a lambda that adds two number
add = lambda a,b:a+b
print(add(10,20))
```

```python
# Use a lambda with map() to convert a list of Celsius temperatures to Fahrenheit.
li = [34,65,78,76,34,21]
cel_to_fah = list(map(lambda a:(a*9/5)+32,li ))
print(cel_to_fah)
```

```python
# Use a lambda with filter() to keep only strings longer than 5 characters from a list.
li = ["hellow","world","abdullah","the","am"]
lg = list(filter(lambda a:len(a)>=5,li))
print(lg)
```

```python
# Use a lambda with sorted() to sort a list of tuples by the second element.
li = [
    (2,1,4,5),
    (6,5,8,9,1),
    (0,4,6,2,1),
    (2,0,4,6,7,8)
]
sorted_li = sorted(li,key=lambda a :a[1])
print(sorted_li)
```

```python
# Write a lambda that takes three arguments and returns the largest.
li = [3,21.4,57,6]
largest = lambda a,b,c:max(a,b,c)
print(largest(3,20,1))
```

```python
# Use a lambda with reduce() (from functools) to compute the product of a list of numbers.
from functools import reduce
li = [1,2,3,4,5]
product = reduce(lambda a,b:a*b,li)
print(product)
```

```python
# Sort a list of dictionaries by a specific key (e.g., 'age') using a lambda as the key argument.
li =[
   { "age" :25},
    {"age" :24},
    {"age" :26},
    {"age" :20},
    ]
sorted_dic = sorted(li,key=lambda a:a["age"])
print(sorted_dic)
```

```python
# What are the limitations of lambda functions compared to regular functions? List at least three.

# Only one expression allowed: Lambda functions can contain only a single expression and cannot include multiple statements.
# No statements like loops or try-except: You cannot directly use statements such as for, while, try, or return inside a lambda.
# Less readable for complex logic: Complex lambda expressions can become difficult to understand and maintain.
# No docstrings: Lambda functions cannot easily include documentation strings like regular functions.
# Harder to debug: Since lambdas are anonymous, error messages may be less descriptive compared to named functions.
```

```python
# Write a lambda-based pipeline that applies a series of transformations to a number: double it, subtract 3, then square it.
pipeline=[
    lambda x:x*2,
    lambda x:x-3,
    lambda x:x**2
]

def apply_pipeline(value,func):
    for i in func:
        value = i(value)
    return value
print(apply_pipeline(5,pipeline))
```

```python
# Use nested lambdas to create a curried addition function: add(3)(4) == 7.
add = lambda x:(lambda y:x+y)
print(add(3)(7))
```

```python
# Implement a lambda-based dispatch table to perform arithmetic operations (+, -, *, /) selected by a string key

operations = {
    "+":lambda a,b: a+b,
    "-":lambda a,b: a-b,
    "*":lambda a,b: a*b,
    "/":lambda a,b: a/b if b!=0 else "Division by zero",
}
print(operations["+"](3,7))
print(operations["*"](3,7))
```

```python
# Write a lambda that flattens one level of nested lists: [[1,2],[3,4]] → [1,2,3,4].
flatten = lambda ist:[item for sublist in ist for item in sublist]
print(flatten([[1,2,3],[4,5,6]]))
```

```python
 # Use sorted() with a multi-key lambda to sort a list of employee dictionaries first by department, then by salary descending.
employees = [
    {"name": "Ali", "department": "IT", "salary": 70000},
    {"name": "Sara", "department": "HR", "salary": 50000},
    {"name": "John", "department": "IT", "salary": 90000},
    {"name": "Emma", "department": "HR", "salary": 60000},
]
sorted_employees = sorted(
    employees,
    key = lambda e:(e["department"],-e["salary"])
)
for emp in sorted_employees:
    print(emp)
```

```python
# Explain when a lambda is preferred over functools.partial. Write an example of each for the same problem.
# When lambda is preferred
# Use a lambda when:
# You need quick inline logic
# Arguments need rearranging or computation
# The transformation is simple and local
# When partial is preferred

# Use functools.partial when:
# You only want to pre-fill arguments
# You want clearer intent
# Reusability/readability matters

from functools import partial
def multiply(a,b):
    return a*b
# lambda
double_lambda = lambda x:multiply(2,x)

# partial
double_partial = partial(multiply,2)

print(double_lambda(2))
print(double_partial(2))
```

```python
# Use a lambda with itertools.groupby() to group a list of words by their length.
from itertools import groupby
words = ["hi", "cat", "dog", "tree", "book", "sun"]
words.sort(key=lambda w:len(w))
grouped={
    length:list(group)
    for length,group in groupby(words,key=lambda w:len(w))
}
print(grouped)
```

```python
# Write code that generates a list of incrementing lambdas using a default argument trick to avoid the late-binding closure problem.
funcs = [lambda x,i=i:x+i for i in range(5)]
for i in funcs:
    print(i(10))
```

```python
# . Demonstrate how to compose two lambdas f and g into a new lambda representing f(g(x)).
f = lambda x:x*2
g = lambda x:x+3
compose = lambda f,g:(lambda x: f(g(x)))
h= compose(f,g)
print(h(5))
```

```python
# Compare the performance of a lambda with a named function doing the same work (time 1 million calls). What do you observe?
import time
lambda_function = lambda x:x*x
def named_function(x):
    return x*x
value = 1_000_000

# lambda time 
start = time.time()
for i in range(value):
    lambda_function(i)
lambda_time = time.time() - start

# named function time
start = time.time()
for i in range(value):
    named_function(i)
named_time = time.time() - start

print("lambda function time", lambda_time)
print("Named function time", named_time)

# Observation
# Performance is usually nearly identical.
# Both are regular function objects internally.
# Named functions may be very slightly faster/readable in practice.
# Choice should usually be based on readability and maintainability, not speed.
```
