```python
# How do you access elements of a list using positive and negative indexing? Give examples.
li=[1,2,3,4,5,6,7,8,9]
print("Positve index")
print(li[0])
print(li[2])
print(li[5])
print("negative index")
print(li[-1])
print(li[-2])
print(li[-5])
```

```python
# What is list slicing? Extract the first three elements, last two elements, and every other element from a list.
li = [1,2,3.4,"abdullah",True]
print("First three element: ",li[0:3])
print("Last Two element: ",li[-1:-3:-1])
```

```python
# Write code to append, insert, and extend a list. What is the difference between append and extend?
li=[10,20,30]
li.append(40) #insert element at last
print(li)
li.insert(1,15) #at specific position
print(li)
li2 = [50,60]
li.extend(li2)
print(li) # merg two lists
```

```python
# How do you remove elements from a list? Show the difference between remove(), pop(), and del.
li=[1,2,3,4,5,6]
li.pop()
print(li) # remove last element
li.remove(2)
print(li) # at specific value
del li[2] # using index
print(li)
li.clear() # delete all entire element
print(li)
```

```python
# Write code to sort a list of numbers in ascending and descending order using sort() and sorted().
li = [2,1,3,4,5,43,6]
li.sort()
print("Ascending using .sort(): ",li)
li.sort(reverse=True)
print("Descending using .sort(): ",li)

# or 
li1 = sorted(li)
print("Ascending using sorted():" ,li1)
li1 = sorted(li,reverse=True)
print("Descending using sorted():" ,li1)
```

```python
# What is list concatenation and repetition? Show examples using + and * operators.
li1 = [1,2,3]
li2=[4,5,6]
li3 = li1+li2 # add two list using concatenation
print(li3)
li3 = li1*4 # add same list by number
print(li3)
```

```python
# Write a program that finds the minimum, maximum, and average of a list of numbers without using built-in min/max.
li = [2,63,4,54]
max=li[0]
min=li[0]
sum = 0
for i in li:
    if max<i:
        max=i
    if min>i:
        min=i
    sum += i
avg = sum/len(li)
print("max: ",max)
print("min: ",min)
print("Avg: ",avg)
```

```python
# How do you check if an item exists in a list? Write an example using 'in' and 'not in'
li = [1,2,3,4,5]
i = 5
if i in li:
    print("Item found: ",i)
j = 10
if j not in li:
    print("Item not found: ",j)
```

```python
# Write a program to reverse a list using slicing, the reverse() method, and the reversed() function.
li = [1,2,3,4,5] 
print("Using slicing [::-1]: ", li[::-1])
li2 = ['a','b','c','d']
li2.reverse()
print("Using reverse(): ", li2)
li3 = [3,2,1,"Tayyab",True]
rev_func = list(reversed(li3))
print("Using reversed function: ", rev_func)
```
