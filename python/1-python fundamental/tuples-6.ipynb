```python
# tuple# What is a tuple in Python? How does it differ from a list? Create a tuple of five colours.
# tuple is immutable and list are mutable
tp = ("black","red","green","white","cyan")
print(tp," : ", type(tp))
```

```python
# What does 'immutable' mean for tuples? Try to change a tuple element and explain the resulting error.
# 'Immutable' means that once a tuple is created, its structure—the elements it contains and their order—cannot be altered. You cannot add, remove, or change elements in place
tp = (1,2,3,4,5)
print(tp[1])
tp[1]=6
print(tp)
```

```python
# What is tuple unpacking? Write code to unpack a tuple (name, age, city) into separate variables.
tp = ("Abdullah",26,"Kasur")
name,age,city = tp
print("name: ",name)
print("age: ",age)
print("city: ",city)
```

```python
# What is the use of a single-element tuple? Write one correctly and one incorrectly, explaining the difference.
# python consider single element with out comma is int value, so create single value tuple is with commma
tp = (10)  # incorrect
print(tp," : ",type(tp))
tp = (20,) # correct
print(tp," : ",type(tp))
```
