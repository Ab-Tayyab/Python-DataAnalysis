```python
# Write a class 'Circle' with an attribute 'radius' and methods area() and circumference()
import math
class Circle:
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return round(math.pi*(self.radius*self.radius),2)
    def circumference(self):
        return round(2*(math.pi*self.radius),2)
measured = Circle(12)
print(measured.area())
print(measured.circumference())
```

```python
# Write a class 'BankAccount' with methods deposit(), withdraw(), and get_balance().
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner  = owner
        self.balance  = balance
    def deposit(self,amount):
        self.balance += amount
        print("You deposit: ",amount)
    def withdraw(self,amount):
        if amount<self.balance:
            self.balance -=amount
            print("You withdraw: ",amount)
        else:
            print("Insufichent Balance")
    def get_balance(self):
        return self.balance
         # User-friendly representation
    def __str__(self):
        return f"Bank Account of {self.owner} with balance {self.balance}"

    # Developer-friendly representation
    def __repr__(self):
        return f"BankAccount(owner='{self.owner}', balance={self.balance})"

bank = BankAccount("Abdullah", 1000)
bank.deposit(500)
bank.withdraw(300)
print("Currnet Balance: ",bank.get_balance())
# Using __str__
print(bank)

# Using __repr__
print(repr(bank))
```
