```python
# class object , method, attribute 
class car:
    def __init__(self,color):
        self.color = color
    def start(self):
        print(f"{self.color} car started")
my_car = car("Black")
my_car.start()
```

```python
# defining class and creating object 
class car:
    def drive(self):
        print("Car is moving!")
myCar = car()
myCar.drive()
```

```python
class car:
    color = "Black"
    def drive(self):
        print(f"{self.color} car is moving")
myCar = car()
myCar.drive()
```

```python
# we can also edit object
class car:
    color = "Black"
    def drive(self):
        print(f"{self.color} car is moving")
    def setColor(self,newColor):
        self.color = newColor
myCar=car()
myCar.setColor("Red")
myCar.drive()
```
