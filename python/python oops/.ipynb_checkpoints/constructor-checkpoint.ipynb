```python
# the __init__() contructor method we call it for setting default value.
class dog:
    def __init__(self):
        self.name = "Buddy"
    def bark(self):
        print(f"{self.name} says woof!")
myDog= dog()
myDog.bark()
```

```python
class car:
    def __init__(self,brand,color):
        self.brand=brand
        self.color=color
myCar=car("Frari","Black")
print(myCar.brand)
print(myCar.color)
myCar.color="Blue"
myCar2=car("Civic","White")
print(myCar2.brand)
print(myCar2.color)
print(myCar.color)
```
