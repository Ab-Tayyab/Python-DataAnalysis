```python
# what is flask
# Flask is lightweight python web framworks
# ideal for building web apps, apis, & ML/DL model deployment
```

```python
# key feature of flask
# minimalist core -> flexible & extensible
# built in development server & debugger
# support templating (jinja2) & routing
```

```python
# why use flask for ML / Dl/AI?
# deploy trained models as web APIs
# serve prediction via HTTP request (rest API)
# simple integration with python Ml libraries: scikit-learn,pyTorch,tensorFlow
# useful for prototypes & lightweight apps: 
# 1-quick testing of models in production-like environments
# 2-expose endpoints for frontend or mobile apps
```

```python
# comparing flask with Django & FastAPI
# Feature          Flask                  Django                      FastAPI
# Complexity       Minimal                Full-featured               Minimal
# Learning curve   easy                   moderate                    easy
# speed            moderate               moderate                    high(async)
# use case         API,lightweight app    full web apps, ORM,Auth     FastAPIs,async ML inferece
# flexibilty       very flexible           less flexible              flexible
```

```python
# Take away
# flask -> simple,flexible,perfect for deploying ML/DL models quickly
# Problem: you trained an ML model -> how does a user use it?
# flask solution: input(user) -> flask API-> ML model-> output (prediction)
```
