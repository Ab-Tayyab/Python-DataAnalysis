```python
# What is a set in Python? Create a set and demonstrate that it stores only unique elements.
sets = {1,2,3,3,4,5,6,6}
print(sets, " : ", type(sets))
```

```python
# Write code to perform union, intersection, difference, and symmetric difference on two sets.
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}
print("Union: A union B: ",set1.union(set2))
print("Intersection: A Intersection B: ",set1.intersection(set2))
print("Difference: A difference B: ",set1-set2)
print("Symmetric difference: A Symmetric difference B: ",set1^set2)
```

```python
# What is a frozenset? When would you use one instead of a regular set?
# A frozenset in Python is an immutable version of a standard set object. Like a regular set, it is an unordered collection of unique elements.However, once a frozenset is created, its contents cannot be modified—you cannot add, remove, or change elements.
fs = ({1,2,3,3,4,5})
print(fs)
```

```python
# Write a program that removes duplicates from a list using a set, preserving the original order.
li = [1,2,2,3,4,5,5,6,7,7,7]
sets= set()
li2 = []
for i in li:
    if i not in sets:
        li2.append(i)
        sets.add(i)
print(sets)
print(li2)
```

```python
# Use set operations to find common friends between two friend lists
li1 = ["Abdullah","Tayyab","Sheraz"]
li2 = ["Tayyab","Sheraz","Ibrahim"]
set1 = set(li1)
set2 = set(li2)
print("Using &: ",set1 & set2) # find using intersection (a & b) or (a intersection b)
print("Using intersection: ",set1.intersection(set2)) # find using intersection (a & b) or (a intersection b)
```
