```python
# What is a lambda function in Python? Write a lambda that adds two number
add = lambda a,b:a+b
print(add(10,20))
```

```python
# Rewrite the function 'def square(x): return x**2' as a lambda expression
sq = lambda a:a**2
print(sq(5))
```

```python
# Use a lambda with map() to convert a list of Celsius temperatures to Fahrenheit.
li = [34,65,78,76,34,21]
cel_to_fah = list(map(lambda a:(a*9/5)+32,li ))
print(cel_to_fah)
```

```python
# Use a lambda with filter() to keep only strings longer than 5 characters from a list.
li = ["hellow","world","abdullah","the","am"]
lg = list(filter(lambda a:len(a)>=5,li))
print(lg)
```

```python
# Use a lambda with sorted() to sort a list of tuples by the second element.
li = [
    (2,1,4,5),
    (6,5,8,9,1),
    (0,4,6,2,1),
    (2,0,4,6,7,8)
]
sorted_li = sorted(li,key=lambda a :a[1])
print(sorted_li)
```

```python
# Write a lambda that takes three arguments and returns the largest.
li = [3,21.4,57,6]
largest = lambda a,b,c:max(a,b,c)
print(largest(3,20,1))
```

```python
# Use a lambda with reduce() (from functools) to compute the product of a list of numbers.
from functools import reduce
li = [1,2,3,4,5]
product = reduce(lambda a,b:a*b,li)
print(product)
```

```python
# Write code that stores multiple lambdas in a dictionary and calls them by name.
dic = {
    "add":lambda a,b:a+b,
    "mul":lambda a,b:a*b,
    "rem":lambda a:a%2
}
print(dic["add"](10,20))
print(dic["mul"](10,20))
print(dic["rem"](10))
```

```python
# Sort a list of dictionaries by a specific key (e.g., 'age') using a lambda as the key argument.
li =[
   { "age" :25},
    {"age" :24},
    {"age" :26},
    {"age" :20},
    ]
sorted_dic = sorted(li,key=lambda a:a["age"])
print(sorted_dic)
```

```python
# What are the limitations of lambda functions compared to regular functions? List at least three.

# Only one expression allowed: Lambda functions can contain only a single expression and cannot include multiple statements.
# No statements like loops or try-except: You cannot directly use statements such as for, while, try, or return inside a lambda.
# Less readable for complex logic: Complex lambda expressions can become difficult to understand and maintain.
# No docstrings: Lambda functions cannot easily include documentation strings like regular functions.
# Harder to debug: Since lambdas are anonymous, error messages may be less descriptive compared to named functions.
```
