```python
# method overriding  use and modified parent class function into child class
class animal:
    def speak(self,a,b):
        print(a+b)
class dog(animal):
    def speak(self,a,b): #redefine function 
        print(a-b)
parentDog=animal()
parentDog.speak(10,20)
childDog = dog()
childDog.speak(10,20)
```
