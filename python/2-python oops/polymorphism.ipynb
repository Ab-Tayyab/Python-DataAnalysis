```python
# poly morphism - overriding % over loading. polymorphism mean multiple , we use diffrent class at a time and call a same function
class Cat:
    def speak(self):
        print("Meo Meo!")
class Dog:
    def speak(self):
        print("Whoo Whoo")
for pets in [Cat(),Dog()]:
    pets.speak()
```

```python
# overriding in polymorphism 
# poly morphism - overriding % over loading. polymorphism mean multiple , we use diffrent class at a time and call a same function
class vehicle:
    def move(self):
        print("Vehicle is moving")
class Car(vehicle):
    def move(self):
        print("Car us driving")
car1 = Car()
car1.move()
```

```python
# method overloading mean we create one function and apply diffrent method as required or diffrent number of arguments
class Calculator:
    def add(self,a,b=0,c=0):
        return a+b+c
cal = Calculator()
print(cal.add(10))
print(cal.add(10,20))
print(cal.add(10,20,30))
```

```python
# speciall method of class that python run by default, it is create by __str__ like constructor
class Book:
    def __init__(self,title,author):
        self.title = title
        self.author = author
    def __str__(self):
        return f"'{self.title}' by {self.author}"
b=Book("Quran","Allah")
print(b)
```

```python
class playList:
    def __init__(self,songs):
        self.songs=songs
    def __len__(self):
        return len(self.songs)
    def __add__(self,other):
        return playList(self.songs + other.songs)
p1 = playList(["Song 1","Song 2"])
p2 = playList(["Song 3"])
print(len(p1))
print(len(p2))
print (len(p1+p2))
```
